```python
# scripts/extract_sqlite_test.py
import sqlite3
import json
from datetime import datetime, timezone

db_path = '../data/leetcode-tracker.sqlite'  # 如果数据库在 data 文件夹
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

now = datetime.now(timezone.utc)
```

```python
cursor.execute("SELECT id, history, tags FROM problems")
rows = cursor.fetchall()
```

```python
for pid, history_json, tags_json in rows:
    if not history_json:
        continue

    try:
        history = json.loads(history_json)
        for h in history:
            dt = datetime.fromisoformat(h["date"].replace("Z", "+00:00"))
            delta = (now - dt).total_seconds() / 3600
            duration = h.get("duration", 0)
            label = 1 if h.get("status") == "Solved" else 0
            tags = json.loads(tags_json or '[]')

            print(f"problem_id: {pid}, label: {label}, tags: {tags}, last_time_diff: {round(delta, 2)}, last duration: {duration}")
            # dataset.append({
            #     "user_id": 0,
            #     "problem_id": pid,
            #     "label": label,
            #     "last_time_diff": round(delta, 2),
            #     "last_duration": duration
            # })
    except Exception as e:
        print(f"❌ 解析失败: {pid} -> {e}")
```
